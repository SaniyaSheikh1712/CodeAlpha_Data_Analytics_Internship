#Load the dataset

In this step, the AI Job Impact dataset is uploaded and loaded into a Pandas DataFrame. The first five records are displayed to verify that the dataset has been imported successfully.

```python
df = pd.read_csv('/content/ai_job_impact.csv')
df.head()
```

# Dataset overview

This step examines the dimensions of the dataset and the names of all available variables. Understanding the number of rows and columns is important before performing data cleaning and analysis.

```python
print('Rows and Columns:', df.shape)

print('\nTotal Rows:', df.shape[0])

print('Total Columns:', df.shape[1])

print('\nColumn Names')

for i, col in enumerate(df.columns, start=1):
    print(f'{i}. {col}')
```

The structure of the dataset is examined using the info() function. This includes data types, total records, and non-null values for each column.

```python
df.info()
```

# Check missing values

This step checks whether any values are missing in the dataset.

```python
missing_values = df.isnull().sum()

print(missing_values)

print('\nTotal Missing Values:', missing_values.sum())
```

# Check duplicate rows

Duplicate records are identified before data cleaning.

```python
duplicates = df.duplicated().sum()

print('Duplicate Rows:', duplicates)
```

# Remove duplicate records

Duplicate records are removed to improve data quality.

```python
df.drop_duplicates(inplace=True)

print('New Shape:', df.shape)
```

# Check unique values

Unique values in categorical columns are examined to understand the available categories and detect inconsistent values.

```python
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    print('\n' + '='*50)
    print('Column:', col)
    print('Unique Values:', df[col].nunique())
    print(df[col].value_counts())
```

# Statistical summary

Descriptive statistics are generated for all numerical columns.

```python
df.describe().T
```

# Categorical summary

A summary of categorical columns is generated.

```python
df.describe(include='object').T
```

# Data cleaning

Extra spaces are removed from all categorical column

```python
for col in categorical_cols:
    df[col] = df[col].str.strip()

print('Text columns cleaned successfully')
```

# Feature engineering

New analytical features are created to measure the impact of AI on employee salaries.

```python
# Salary Difference
df['Salary_Difference'] = (
    df['Salary_After_AI'] -
    df['Salary_Before_AI']
)

# Salary Growth Percentage
df['Salary_Growth_%'] = (
    df['Salary_Difference'] /
    df['Salary_Before_AI']
) * 100

df[['Salary_Before_AI',
    'Salary_After_AI',
    'Salary_Difference',
    'Salary_Growth_%']].head()
```

# Create experience categories

Employees are grouped into experience categories for easier analysis.

```python
def experience_category(exp):

    if exp <= 2:
        return 'Fresher'

    elif exp <= 5:
        return 'Junior'

    elif exp <= 10:
        return 'Mid-Level'

    else:
        return 'Senior'

df['Experience_Category'] = (
    df['Years_Experience']
    .apply(experience_category)
)

df[['Years_Experience',
    'Experience_Category']].head()
```

# Save cleaned dataset

The cleaned and transformed dataset is exported as a new CSV file for further analysis.

```python
df.to_csv(
    'AI_Job_Impact_Cleaned.csv',
    index=False
)

print('Cleaned dataset saved successfully')
```
