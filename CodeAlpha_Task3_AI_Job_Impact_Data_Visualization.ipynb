### AI Job Impact Analysis

### Objective

The objective of this task is to transform the AI Job Impact dataset into meaningful visualizations. The charts help identify trends, patterns, and relationships between AI adoption, salaries, automation risk, productivity, and job satisfaction.

The visualizations are created using **Matplotlib** and **Seaborn** in Google Colab.


# IMPORT LIBRARIES

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
df = pd.read_csv('/content/ai_job_impact.csv')

# Chart settings
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10,6)

print('Dataset loaded successfully')
print(df.shape)
```

#  Chart 1: Age distribution

This histogram shows the distribution of employee ages in the dataset. It helps identify which age groups are most represented.


```python
plt.figure(figsize=(8,5))

sns.histplot(df['Age'], bins=15, kde=True,color="orange")

plt.title('Age Distribution of Employees')
plt.xlabel('Age')
plt.ylabel('Number of Employees')

plt.savefig('age_distribution.png', dpi=300, bbox_inches='tight')

plt.show()
```

# Chart 2: Gender distribution

This count plot displays the number of male and female employees in the dataset.


```python
plt.figure(figsize=(6,5))

sns.countplot(x='Gender', data=df)

plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')

plt.savefig('gender_distribution.png', dpi=300, bbox_inches='tight')

plt.show()
```

#Chart 3: Education level distribution

This chart shows the distribution of employees across different education levels.


```python
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA726', '#66BB6A']

plt.figure(figsize=(8,5))

sns.countplot(
    x='Education_Level',
    data=df,
    palette=colors
)

plt.title('Education Level Distribution')
plt.xlabel('Education Level')
plt.ylabel('Count')
plt.xticks(rotation=20)

plt.savefig('education_distribution.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Salary_Before_AI'], bins=20, kde=True,color="greenyellow")

plt.title('Salary Distribution Before AI')
plt.xlabel('Salary Before AI')
plt.ylabel('Number of Employees')

plt.savefig('salary_before_ai.png', dpi=300, bbox_inches='tight')

plt.show()
```
